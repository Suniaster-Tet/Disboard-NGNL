# Disboard-NGNL — Notebook de demostración

Este notebook muestra cómo invocar los componentes principales del ecosistema
desde Python, sin pasar por la consola ni por el servidor web.

> **Requisito:** haber ejecutado `make install` o `pip install -e ".[dev]"`.

```python
# 1. Crear un edge de ejemplo
from pathlib import Path
from core.edge import Edge

seed = Path("examples/edge-example/seed.md")
edge = Edge.crear_desde_semilla(seed, titulo="Demo notebook")
print(f"Edge creado: {edge.id}")
print(f"Ruta: {edge.ruta}")
```

```python
# 2. Sora propone el plan
from agents.sora.agent import SoraStrategist

sora = SoraStrategist(edge_id=edge.id)
plan = sora.plan_edge(edge.id, seed.read_text(encoding="utf-8"))
for obj in plan["objetivos"]:
    print(f"- {obj}")
```

```python
# 3. Suniaster enruta un modelo para Sora
from suniaster.router import SuniasterRouter

router = SuniasterRouter()
model = router.get_model_for("sora")
print(f"Modelo asignado a Sora: {model}")
```

## Notas

- Si `WOLFRAM_API_KEY` no está configurada, Suniaster asigna a Shiro un
  proveedor alternativo y Shiro declara qué proveedor usó; Tet valida.
- Si `TET_SIGNING_KEY` no está configurada, las funciones de HITL lanzan
  `RuntimeError`; configura `.env` antes de ejecutar el flujo completo.
